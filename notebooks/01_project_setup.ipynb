```python
import pandas as pd
import numpy as np
import sys
import pickle
import gc

# 1. Load dataset penuh (811K baris) - HATI-HATI MEMORI
print("Loading full dataset...")
# Fix kompatibilitas pandas versi lama/baru
sys.modules['pandas.indexes'] = pd.core.indexes 

file_path = '../data/LSWMD.pkl'
with open(file_path, 'rb') as f:
    df_full = pickle.load(f, encoding='latin1') # Dataframe disimpan di df_full
    
print(f"Data berhasil di load, bentuk data: {df_full.shape}")

# 2. Buat fungsi extract_label yang AMAN dari error TypeError
def extract_label(val):
    # Cek jika val adalah list/array dan tidak kosong
    if isinstance(val, (list, np.ndarray)) and len(val) > 0:
        inner = val[0]
        if isinstance(inner, (list, np.ndarray)) and len(inner) > 0:
            return inner[0]
    return 'none'

# 3. Terapkan ke df_full (BUKAN Tempat_data) untuk mendapatkan label flat
print("Extracting labels...")
df_full['ft_flat'] = df_full['failureType'].apply(extract_label)

# 4. Filter baris yang berlabel Donut dan Near-full
print("Filtering rare defects...")
donut_df = df_full[df_full['ft_flat'] == 'Donut'].head(1)
nearfull_df = df_full[df_full['ft_flat'] == 'Near-full'].head(2)

print(f"Donut found: {len(donut_df)}")
print(f"Near-full found: {len(nearfull_df)}")

# 5. Load sample lama (1000 baris) yang sudah kamu simpan di Day 11
df_sample = pd.read_pickle('../data/wm811k_sample.pkl')

# 6. Gabungkan sample lama dengan baris langka
# Kita drop kolom ft_flat dari donut/nearfull agar tidak ikut ke df_sample
donut_df = donut_df.drop(columns=['ft_flat'])
nearfull_df = nearfull_df.drop(columns=['ft_flat'])

df_sample = pd.concat([df_sample, donut_df, nearfull_df], ignore_index=True)

# 7. Hapus df_full dari memori agar laptop tidak nge-hang
del df_full
gc.collect()

print("\nSampling selesai. Shape sample sekarang:", df_sample.shape)

# 8. Verifikasi 8 kategori di sample baru
df_sample['failureType_clean'] = df_sample['failureType'].apply(extract_label)
print("\nDistribusi failureType_clean setelah digabung:")
print(df_sample['failureType_clean'].value_counts())

# 9. Save/Overwrite sample baru
df_sample.to_pickle('../data/wm811k_sample.pkl')
print("\nSample baru berhasil disimpan ke ../data/wm811k_sample.pkl")
```

```python
# ============================================================
# Day 12 — Stratified Random Sampling dengan Composite Key (v2)
# Preview teknik groupby + apply (McKinney Bab 10, bahan Day 13)
# ============================================================
import pandas as pd
import numpy as np
import sys
import pickle
import gc

# ============================================================
# KONFIGURASI — satu angka CAP ini yang di-tuning, bukan per-kategori manual
# ============================================================
CAP =  638507
RANDOM_STATE = 42
OUTPUT_PATH = '../data/wm811k_sample_v2_stratified.pkl'

# ============================================================
# 1. Load dataset penuh
# ============================================================
print("Loading full dataset...")
sys.modules['pandas.indexes'] = pd.core.indexes

file_path = '../data/LSWMD.pkl'
with open(file_path, 'rb') as f:
    df_full = pickle.load(f, encoding='latin1')

print(f"Data berhasil di load, bentuk data: {df_full.shape}")

# ============================================================
# 2. Extract label bersih — DUA-DUANYA sekarang: failureType DAN
#    trianTestLabel, karena composite key butuh keduanya
# ============================================================
def extract_label(val):
    if isinstance(val, (list, np.ndarray)) and len(val) > 0:
        inner = val[0]
        if isinstance(inner, (list, np.ndarray)) and len(inner) > 0:
            return inner[0]
    return 'none'

print("Extracting labels...")
df_full['failureType_clean'] = df_full['failureType'].apply(extract_label)
df_full['trainTestLabel_clean'] = df_full['trianTestLabel'].apply(extract_label)

# ============================================================
# 3. Composite stratify key — pisahin "belum pernah direview" vs
#    "sudah direview, hasilnya bersih", biar insight yang sudah
#    kamu validasi gak ketutup jadi satu stratum "none" gado-gado.
#    np.where (vectorized) dipakai, bukan .apply(axis=1), karena
#    row-wise apply lambat banget di 811K baris.
# ============================================================
df_full['stratify_key'] = np.where(
    df_full['trainTestLabel_clean'] == 'none',
    'unlabeled',
    np.where(df_full['failureType_clean'] == 'none', 'reviewed_no_defect', df_full['failureType_clean'])
)

print("\nDistribusi stratify_key di FULL dataset:")
print(df_full['stratify_key'].value_counts())

# ============================================================
# 4. Sampling generik: loop eksplisit per grup + cap + random sample
#    min(len(group), cap) otomatis handle kategori yang populasinya
#    di bawah cap (misal Near-full) tanpa exception khusus.
#    Loop eksplisit dipakai (bukan groupby().apply()) karena versi
#    pandas terbaru mengecualikan kolom grouping dari data yang
#    diterima fungsi di dalam .apply() — bikin 'stratify_key' hilang
#    dari hasil. Iterasi manual gak punya masalah ini di versi manapun.
# ============================================================
sampled_groups = []
for key, group in df_full.groupby('stratify_key'):
    n = min(len(group), CAP)
    sampled_groups.append(group.sample(n=n, random_state=RANDOM_STATE))

df_stratified = pd.concat(sampled_groups, ignore_index=True)

print("\nDistribusi stratify_key setelah stratified sampling:")
print(df_stratified['stratify_key'].value_counts())
print(f"\nShape akhir: {df_stratified.shape}")

# ============================================================
# 5. Save — wm811k_sample.pkl (v1) sama sekali gak disentuh
# ============================================================
df_stratified.to_pickle(OUTPUT_PATH)
print(f"\nSample v2 berhasil disimpan ke {OUTPUT_PATH}")

# ============================================================
# 6. Bersihkan df_full dari memori
# ============================================================
del df_full
gc.collect()
```
